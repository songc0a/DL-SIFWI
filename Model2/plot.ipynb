```python
import numpy as np
from scipy.ndimage import gaussian_filter
import matplotlib.pyplot as plt
```

```python
train_vel = []
for i in range(1, 41):

    vel = np.load('./model%d.npy'%(i))
#     vel = np.load('./Model1/model%d.npy'%(i))
    vels = gaussian_filter(vel, sigma=2)
    train_vel.append(vels)
train_vel = np.array(train_vel)
print(train_vel.shape)
```

```python
from mpl_toolkits.axes_grid1 import make_axes_locatable
plt.style.use('default')

plt.figure(figsize=(19,4))

plt.subplot(1,4,1)
ax = plt.gca()
im = ax.imshow(train_vel[11,:,:],  aspect=1, cmap="jet")
plt.xlabel('Distance (km)', fontsize=14)
plt.xticks(fontsize=10)

plt.ylabel('Depth (km)', fontsize=14)
plt.yticks(fontsize=10)
ax.yaxis.set_major_locator(plt.MultipleLocator(0.5))

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="6%", pad=0.15)
cbar = plt.colorbar(im, cax=cax)
plt.subplot(1,4,2)
ax = plt.gca()
im = ax.imshow(train_vel[12,:,:],  aspect=1, cmap="jet")

plt.xlabel('Distance (km)', fontsize=14)
plt.xticks(fontsize=10)

ax.yaxis.set_major_locator(plt.MultipleLocator(0.5))

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="6%", pad=0.15)

cbar = plt.colorbar(im, cax=cax)
plt.subplot(1,4,3)
ax = plt.gca()
im = ax.imshow(train_vel[2,:,:],  aspect=1, cmap="jet")


plt.xlabel('Distance (km)', fontsize=14)
plt.xticks(fontsize=10)

ax.yaxis.set_major_locator(plt.MultipleLocator(0.5))

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="6%", pad=0.15)

cbar = plt.colorbar(im, cax=cax)
plt.subplot(1,4,4)
ax = plt.gca()
im = ax.imshow(train_vel[5,:,:], aspect=1, cmap="jet")


plt.xlabel('Distance (km)', fontsize=14)
plt.xticks(fontsize=10)

ax.yaxis.set_major_locator(plt.MultipleLocator(0.5))

divider = make_axes_locatable(ax)
cax = divider.append_axes("right", size="6%", pad=0.15)

cbar = plt.colorbar(im, cax=cax)
```
